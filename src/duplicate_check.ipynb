```python
import pandas as pd
from data_setup import FEATURES, TARGET

df = pd.read_csv("../processed_data/cleaned_fuel.csv")
key = ['make', 'model'] + FEATURES + [TARGET]

is_removed = df.duplicated(subset=key, keep="first")

removed = df[is_removed]
kept = df[~is_removed]

print("Rows removed:", len(removed))
print("Rows kept:", len(kept))
```

```python
df["group_size"] = df.groupby(key, dropna=False)[TARGET].transform("size")

groups = df[df["group_size"] > 1].sort_values(key)

cols = ["year", "make", "model", "engine_descriptor"] + key[1:] + ["group_size"]
groups[cols].head(30)
```

```python
print("Total rows removed:", len(removed))
print("Total rows kept:", len(kept))
```
